```python
file = open("example.txt", "r")  # Open the file in read mode
content = file.read()  # Read the entire content of the file
print("File content:", content)  # Print the content
file.close()  # Close the file  
```

```python
file = open("example.txt", "w") 
file.write("This is a new line of text added----------.")  # Write a new line to the file
file.close()  # Close the file
```

```python
file = open("example.txt", "a")  # Open the file in append mode
file.write("\nThis is an appended line.")  # Append a new line to the file
file .write("\nThis is another appended line.")  # Append another new line to the file
file.close()  # Close the file
```

```python
with open("example.txt", "r") as file:  # Open the file in read mode using 'with' statement
    content = file.read()  # Read the entire content of the file
    print("File content after writing and appending:", content)  # Print the content
```
